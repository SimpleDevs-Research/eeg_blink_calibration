# EEG Blink Calibration

This is a collection of scripts meant to aid in calibrating eye tracking to eeg data. Both have records of blink data, so we can perhaps find a way to align them. This is a bigger issue than normal because **EEG and eye tracking data are, generally, noisy**.

## Hardware & Software Involved

We are using the following devices

- **Unity3D (U3)**: The simulation environment. Runs everything VR-related.
- **Meta Quest Pro (MQP)**: The VR head-mounted display (HMD). Comes with an in-built eye-tracker.
- **Muse S (MS)**: The EEG headband. Has a sampling frequency of 255Hz.
- **Mind Monitor (MM)**: The data handler that acts as a middleman and data aggregator for the EEG headband.

Our analysis is expected to be done via Python, for simplicity.

## Folder Structure

This project is divided in the following manner:

- `src/`: Where relevant notebooks are stored.
    - `data/`: Where participant data is stored. Is not stored locally in the repo; is only accessible to lab members.
    - `components/`: Where helper functions and scripts are stored. Can be used across multiple scripts.
        - `timestamps.py`: Helper functions to transform timestamp data.
        - `file_handling.py`: Helper functions to do OS operatiosn such as creating new directories.
    - `separate.ipynb`: First major notebook file. Handles all data divisions and separations into separate directories.
- `README.md`: For Github; simply describes what each file is meant to do.
- `README.ipynb`: This file!
